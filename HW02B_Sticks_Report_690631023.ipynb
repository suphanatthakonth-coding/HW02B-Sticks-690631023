## 1. Background and Game Rules

Sticks in the Pile is a two-player, turn-based game. There is a pile of sticks on the table. Players take turns removing sticks from the pile, and whoever is forced to take the last stick loses the game.

**Rules :**

- The pile starts with a number of sticks n. In Homework 02B the number is asked inside the program.
- There are always exactly two players: **Player 1 (COMPUTER-AI)** and **Player 2 (USER)**.
- Players take turns.
- On each turn, the current player removes either **1 or 2** sticks. No other amount is allowed.
- A player may not remove more sticks than are currently left in the pile. For example, if only 1 stick remains, the player cannot remove 2.
- The game ends when there are **0** sticks left in the pile.
- The player who removes the last stick (the one who makes the pile reach 0) is the **LOSER**. The other player is the **WINNER**.

**Coding constraints:** only `input()`, `print()`, variables, `if / elif / else`, and a loop (`while` or `for`) are used. There is no `def`, no list or dictionary, and no `import`.

## 2. Theory: Why the Computer Can Play Smartly

**Step-by-step analysis (the player to move can remove 1 or 2 sticks):**

| Sticks in the pile | Result for the player to move | Reason |
|:---:|:---:|:---|
| 1 | Losing | The player must take the last stick. |
| 2 | Winning | Remove 1 stick and leave 1 stick for the opponent (a losing position). |
| 3 | Winning | Remove 2 sticks and leave 1 stick for the opponent (a losing position). |
| 4 | Losing | Any move leaves 2 or 3 sticks, and both are winning positions for the opponent. |
| 5 | Winning | Remove 1 stick and leave 4 sticks (a losing position). |
| 6 | Winning | Remove 2 sticks and leave 4 sticks (a losing position). |
| 7 | Losing | Any move leaves 5 or 6 sticks, and both are winning positions for the opponent. |
| 8 | Winning | Remove 1 stick and leave 7 sticks (a losing position). |
| 9 | Winning | Remove 2 sticks and leave 7 sticks (a losing position). |
| 10 | Losing | Any move leaves 8 or 9 sticks, and both are winning positions for the opponent. |

**Pattern:** the losing positions are 1, 4, 7, 10, ... The pattern repeats every 3 sticks, because a player can remove only 1 or 2 sticks, and whatever one player removes (1 or 2), the other player can remove the remaining part of a group of 3 (2 or 1).

**Remainder (modulo) operator `%`:** `a % b` gives the remainder when `a` is divided by `b`. For example, `10 % 3 = 1` and `9 % 3 = 0`. A pile is a losing position exactly when `sticks % 3 == 1`.

**Strategy for the computer:** always leave the user a losing position (a pile with remainder 1 when divided by 3).

| `sticks % 3` (AI's turn) | Position for the AI | AI removes | Sticks left for the user |
|:---:|:---:|:---:|:---|
| 0 | Winning | 2 | `sticks - 2`, remainder 1 |
| 2 | Winning | 1 | `sticks - 1`, remainder 1 |
| 1 | Losing (against a perfect player) | 1 | `sticks - 1`, remainder 0. The AI waits for the user to make a mistake. |

In code this becomes: **if `sticks % 3 == 0`, remove 2; otherwise remove 1.** This rule works for any starting number n, not only 10.

**Note:** the AI is Player 1 and moves first. If `n % 3 != 1`, the AI starts in a winning position. If `n % 3 == 1` (for example n = 10), the AI starts in a losing position, and **a perfect user can win**.

## 4. Flowchart of the Program

![Flowchart of the Sticks in the Pile AI game](flowchart.png)

**How to read it (top to bottom):**

1. START, then set `player1 = "AI"` and read `player2`.

3. Set `current_player = 1`, so the AI moves first.
4. Loop while `sticks > 0`. Show the sticks remaining.

## 5. The Program

The program is split into four code cells.

### Cell 1 – Setup: player names 

Set the name of Player 1 and ask for the name of Player 2.

**Variables defined:**
- `player1` – a string fixed to `"AI"`, because Player 1 is always the computer, so the program does not ask for it.
- `player2` – a string read with `input()`, because the name of Player 2 must be asked for and stored.

```python
player1 = "AI"  # Player 1 is always the computer, so its name is fixed
print("Enter name for Player 1:", player1)  # show Player 1's name (AI)
player2 = input("Enter name for Player 2: ")  # ask for the user's name
print("The name for Player 2 is", player2)  # confirm Player 2's name
```

### Cell 2 – Setup: starting number of sticks 

Ask the user for the starting number of sticks n and store it in `sticks`. If the value is smaller than 1, it is rejected and asked again.

**Variables defined:**
- `sticks` – an integer. `input()` returns text, so `int()` converts it to a number that can be compared and subtracted.


```python
sticks = int(input("Enter the starting number of sticks: "))  # ask for n
while sticks < 1:  # n must be at least 1, so repeat while it is smaller
    print("Invalid number. Please enter 1 or more.")  # reject the value
    sticks = int(input("Enter the starting number of sticks: "))  # ask again
```

### Cell 3 – Turn control (Section 3.2)

Decide who moves first.

**Variables defined:**
- `current_player` – an integer. `1` means Player 1 (AI) and `2` means Player 2 (user).


```python
current_player = 1  # 1 = Player 1 (AI), 2 = Player 2 (user); AI starts
```

### Cell 4 – Main game loop (Sections 3.2 to 3.5)

Repeat the turns until the pile reaches 0. Each pass through the loop is one turn.

**Variables used:**
- `sticks`, `current_player`, `player1`, `player2` – created in the cells above.
- `remove` – **defined in this cell**. It has no value before the first turn and gets a new value in every turn: from the AI rule or from the user's validated input.

```python
while sticks > 0:  # keep playing while there are sticks in the pile
    print("Sticks remaining:", sticks)  # show the current pile size

    if current_player == 1:  # it is Player 1's (the AI's) turn
        print(player1 + "'s turn.")  # show whose turn it is
        if sticks % 3 == 0:  # remainder 0 -> take 2, leaving remainder 1
            remove = 2  # AI removes 2 sticks
        else:  # remainder 1 or 2 -> take 1
            remove = 1  # AI removes 1 stick (for remainder 2, leaves 1)
        print(player1, "removes", remove, "stick(s).")  # show AI's move
    else:  # it is Player 2's (the user's) turn
        print(player2 + "'s turn.")  # show whose turn it is
        remove = int(input(  # ask the user how many sticks to remove
            "Remove how many sticks (1 or 2)? "))  # prompt text
        while remove < 1 or remove > 2 or remove > sticks:  # invalid amount?
            print("Invalid amount. Please enter 1 or 2.")  # reject the move
            print(player2 + "'s turn.")  # same player's turn again
            remove = int(input(  # ask the same player again
                "Remove how many sticks (1 or 2)? "))  # prompt text

    sticks = sticks - remove  # update the pile after a valid move

    if sticks == 0:  # the pile is empty: the game is over
        print("Sticks remaining:", sticks)  # show the final pile (0)
        if current_player == 1:  # Player 1 took the last stick
            print(player1, "took the last stick.")  # Player 1 is the loser
            print("Winner:", player2)  # so Player 2 is the winner
        else:  # Player 2 took the last stick
            print(player2, "took the last stick.")  # Player 2 is the loser
            print("Winner:", player1)  # so Player 1 is the winner
    else:  # the game is not over yet
        if current_player == 1:  # Player 1 has just moved
            current_player = 2  # next turn belongs to Player 2
        else:  # Player 2 has just moved
            current_player = 1  # next turn belongs to Player 1
```

## 6. Written Explanation

**How I kept track of whose turn it was.** I use one variable, `current_player`. It is set to 1 before the loop starts, so Player 1 (the AI) moves first. Inside the loop, an `if / else` on `current_player` chooses between the AI's turn and the user's turn. After a valid move that does not end the game, the program switches the value (1 becomes 2, and 2 becomes 1). The switch happens only after a valid move, so the two players always alternate correctly.

**How I made the computer smart.** I analysed the game backwards from the end (Section 2). A pile of 1 stick is a losing position for the player to move, and the losing positions repeat every 3 sticks: 1, 4, 7, 10, ... These are the piles where `sticks % 3 == 1`. The computer therefore always tries to leave the user with a pile that has remainder 1 when divided by 3. It removes 2 sticks when `sticks % 3 == 0` and 1 stick otherwise. This works for any starting number n. If the AI is already in a losing position, it removes 1 stick and waits for the user to make a mistake.

**How the winner is decided.** After each valid move the pile is updated. When `sticks == 0`, the player stored in `current_player` made the pile reach 0, so that player took the last stick and is the loser. The other player is printed as the winner.
